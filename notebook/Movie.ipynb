```python
df=df.drop_duplicates().reset_index(drop=True)
```

```python
df = df[["title","overview","genres","tagline","vote_average","popularity"]]
```

```python
df = df.dropna(subset=["title"])
```

```python
df["title"] = df["title"].str.lower()
```

```python
df["overview"] = df["overview"].fillna("")
```

```python
df.iloc[0]["genres"]
```

```python
import ast
df["genres"] = df["genres"].apply(lambda x:" ".join([i["name"] for i in ast.literal_eval(x)]))
```

```python
df["tagline"] = df["tagline"].fillna("")
```

```python
df["tages"] = df["overview"] + " " + df["genres"] + " " + df["tagline"]
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import nltk.tokenize
import re

nltk.download('stopwords')
nltk.download('wordnet')
```

```python
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
```

```python
def preprocess_text(text):
  #Lower case
  text = str(text).lower()
  #remove panchuation
  text = re.sub(r'[^a-zA-Z\s]','',text)
  words = text.split()
  #remove stopwords
  words = [word for word in words if word not in stop_words]

  #lemmatize
  words = [lemmatizer.lemmatize(word) for word in words]

  return " ".join(words)

```

```python
df["tages"] = df["tages"].apply(preprocess_text)
```

```python
df = df.reset_index(drop=True)
```

```python
indices = pd.Series(df.index,index=df["title"]).drop_duplicates()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

```

```python
tfidf = TfidfVectorizer(max_features=50000,ngram_range=(1,2),stop_words="english")
```

```python
tfidf
```

```python
tfidf_matrix = tfidf.fit_transform(df['tages'].fillna(''))
```

```python
tfidf_matrix
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
from sklearn.metrics.pairwise import cosine_similarity
import difflib

def recommend(title, n=10):

    title = title.lower().strip()

    
    close_matches = difflib.get_close_matches(
        title,
        df['title'].tolist(),
        n=1,
        cutoff=0.4
    )

    if not close_matches:
        return "Movie not found"

    movie_name = close_matches[0]

    print("Matched Movie:", movie_name)

    idx = indices[movie_name]

    sim_scores = cosine_similarity(
        tfidf_matrix[idx],
        tfidf_matrix
    ).flatten()

    movie_indices = sim_scores.argsort()[::-1][1:n+1]

    return df['title'].iloc[movie_indices]
```
